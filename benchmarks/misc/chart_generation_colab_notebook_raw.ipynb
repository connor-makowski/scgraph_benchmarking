```python
import pandas as pd
import numpy as np
from sklearn.metrics import r2_score



# True and predictions
y_true = df['test_google_length_km']
y_pred_us = df['test_us_freeway_scgraph_length_km']
y_pred_world = df['test_world_highways_scgraph_length_km']

# Compute metrics
metrics = {
    'Comparison': ['US Freeway vs Google', 'World Highways vs Google'],
    'R²': [
        r2_score(y_true, y_pred_us),
        r2_score(y_true, y_pred_world)
    ],
    'MAPE (%)': [
        np.mean(np.abs((y_true - y_pred_us) / y_true)) * 100,
        np.mean(np.abs((y_true - y_pred_world) / y_true)) * 100
    ]
}

metrics_df = pd.DataFrame(metrics)

# Display metrics table
display(metrics_df)
# Prepare summary statistics
stats = []
for label, col in [
    ('US Freeway', 'test_us_freeway_scgraph_time_ms'),
    ('World Highways', 'test_world_highways_scgraph_time_ms')
]:
    stats.append({
        'Comparison': f'{label} vs Google',
        'SCGraph Mean Time (ms)': df[col].mean(),
        'SCGraph Std Time (ms)': df[col].std(),
        'Google Mean Time (ms)': df['test_google_time_ms'].mean(),
        'Google Std Time (ms)': df['test_google_time_ms'].std()
    })

stats_df = pd.DataFrame(stats)

# Display the table
display(stats_df)
pd.merge(metrics_df, stats_df, on='Comparison')
```

```python
import pandas as pd
import folium
import ast
from scgraph.geographs.us_freeway import us_freeway_geograph

# Utility functions to avoid antimeridian clipping in Folium
def adjustArcPath(path):
    for i in range(1, len(path)):
        lon, prev_lon = path[i][1], path[i-1][1]
        path[i][1] = lon - round((lon - prev_lon) / 360, 0) * 360
    return path

def modifyArcPathLong(points, offset):
    return [[lat, lon + offset] for lat, lon in points]

def getCleanArcPath(path):
    path = adjustArcPath(path.copy())
    return [
        path,
        modifyArcPathLong(path, +360),
        modifyArcPathLong(path, -360),
        modifyArcPathLong(path, +720),
        modifyArcPathLong(path, -720),
    ]

# Load your origin–destination pairs
df = pd.read_csv('comparison_tests.csv')
df['coord1'] = df['coord1'].apply(ast.literal_eval)
df['coord2'] = df['coord2'].apply(ast.literal_eval)

# Initialize a Folium map centered on the continental US
m = folium.Map(
    location=[39.0, -98.0],
    zoom_start=4,
    tiles='CartoDB Positron'
)

# Fetch and plot the actual SCGraph US Freeway routes
for idx, row in df.iterrows():
    origin = {"latitude": row['coord1'][0], "longitude": row['coord1'][1]}
    dest   = {"latitude": row['coord2'][0], "longitude": row['coord2'][1]}

    freeway_output = us_freeway_geograph.get_shortest_path(
        origin_node=origin,
        destination_node=dest,
        output_units='mi'
    )
    raw_path = freeway_output['coordinate_path']

    for segment in getCleanArcPath(raw_path):
        folium.PolyLine(
            locations=segment,
            color='blue',
            weight=1,
            opacity=0.4
        ).add_to(m)

# Save to an HTML file (or display `m` in a Jupyter notebook)
m#m.save('us_od_scgraph_map.html')
```

```python
import json, math, os, folium, matplotlib.pyplot as plt

# Paths to uploaded geojson files
oak_path = 'oak_ridge_maritime.geojson'
marnet_path = 'marnet.geojson'
world_highways_path = 'world_highways.geojson'
world_highways_and_marnet_path = 'world_highways_and_marnet.geojson'
world_railways_path = 'world_railways.geojson'


# ---------- Helper functions ----------
def extract_line_coords(path):
    """Extracts lists of (lon, lat) sequences from a GeoJSON file."""
    with open(path, 'r') as f:
        data = json.load(f)

    lines = []

    def process_geom(geom):
        t = geom['type']
        if t == 'LineString':
            lines.append(geom['coordinates'])
        elif t == 'MultiLineString':
            lines.extend(geom['coordinates'])
        elif t == 'Polygon':
            lines.extend(geom['coordinates'])
        elif t == 'MultiPolygon':
            for poly in geom['coordinates']:
                lines.extend(poly)
        elif t == 'GeometryCollection':
            for g in geom['geometries']:
                process_geom(g)
        elif t == 'Feature':
            process_geom(geom['geometry'])

    if data.get('type') == 'FeatureCollection':
        for feat in data['features']:
            process_geom(feat['geometry'])
    else:
        process_geom(data)
    return lines

def geojson_bounds(paths):
    """Compute combined bounds [[minLat,minLon],[maxLat,maxLon]] for a list of GeoJSON files."""
    minlon = math.inf
    minlat = math.inf
    maxlon = -math.inf
    maxlat = -math.inf

    def update_bounds(coords):
        nonlocal minlon, minlat, maxlon, maxlat
        if isinstance(coords, (list, tuple)) and len(coords) > 0 and isinstance(coords[0], (float, int)):
            lon, lat = coords[0], coords[1]
            minlon = min(minlon, lon)
            minlat = min(minlat, lat)
            maxlon = max(maxlon, lon)
            maxlat = max(maxlat, lat)
        elif isinstance(coords, (list, tuple)):
            for c in coords:
                update_bounds(c)

    for path in paths:
        with open(path, 'r') as f:
            data = json.load(f)

        def recurse(obj):
            t = obj.get('type')
            if t in ('LineString', 'MultiLineString', 'Polygon', 'MultiPolygon'):
                update_bounds(obj['coordinates'])
            elif t == 'GeometryCollection':
                for g in obj['geometries']:
                    recurse(g)
            elif t == 'Feature':
                recurse(obj['geometry'])
            elif t == 'FeatureCollection':
                for feat in obj['features']:
                    recurse(feat['geometry'])

        recurse(data)

    return [[minlat, minlon], [maxlat, maxlon]]

def create_folium_map(geojson_path, bounds, html_out):
    """Create and save a Folium map."""
    center = [(bounds[0][0] + bounds[1][0]) / 2, (bounds[0][1] + bounds[1][1]) / 2]
    m = folium.Map(location=center, zoom_start=2, width=800, height=600)
    m.fit_bounds(bounds)
    folium.GeoJson(geojson_path, name=os.path.basename(geojson_path)).add_to(m)
    m

def create_static_pdf(lines, bounds, pdf_out, color='blue', linewidth=0.2, max_lines=10000):
    """Create a static PDF snapshot of the network."""
    if max_lines and len(lines) > max_lines:
        lines = lines[:max_lines]  # simple subsample for performance

    fig, ax = plt.subplots(figsize=(11, 6))
    for ln in lines:
        if len(ln) < 2:
            continue
        lons, lats = zip(*ln)
        ax.plot(lons, lats, color=color, linewidth=linewidth)

    ax.set_xlim(bounds[0][1], bounds[1][1])
    ax.set_ylim(bounds[0][0], bounds[1][0])
    ax.set_xlabel('Longitude')
    ax.set_ylabel('Latitude')
    ax.set_aspect('equal', adjustable='box')
    ax.grid(True, linestyle='--', linewidth=0.3)
    fig.savefig(pdf_out, bbox_inches='tight')
    plt.close(fig)

# ---------- Process the files ----------

# 1. Extract line coordinates
lines_oak = extract_line_coords(oak_path)
lines_marnet = extract_line_coords(marnet_path)
lines_world_highways_path = extract_line_coords(world_highways_path)
lines_world_highways_and_marnet_path = extract_line_coords(world_highways_and_marnet_path)
lines_world_railways_path = extract_line_coords(world_railways_path)



# 2. Compute common map bounds so both maps have the same global frame
common_bounds = geojson_bounds([oak_path, marnet_path])
#common_bounds = geojson_bounds([oak_path, marnet_path, world_highways_path, world_highways_and_marnet_path, world_railways_path])




# print("✓ Created Folium HTML maps and corresponding PDF snapshots.")
# print(f"  - Oak Ridge Maritime HTML: {oak_html}")
# print(f"  - Oak Ridge Maritime PDF : {oak_pdf}")
# print(f"  - Marnet HTML            : {marnet_html}")
# print(f"  - Marnet PDF             : {marnet_pdf}")
```

```python
import osmnx as ox
import networkx as nx
import math, json, requests
from time import time
from pamda import pamda

from scgraph_data.world_highways import world_highways_geograph
from scgraph.geographs.us_freeway import us_freeway_geograph

ox.settings.log_console = True
ox.settings.use_cache = True

# with open('keys.json', 'r') as f:
#     keys = json.load(f)
#     google_api_key = keys.get('google_api_key')

def haversine_distance(coord1, coord2):
    # Calculate great-circle distance in meters
    R = 6371000  # Earth radius in meters
    lat1, lon1 = coord1
    lat2, lon2 = coord2
    phi1, phi2 = math.radians(lat1), math.radians(lat2)
    dphi = math.radians(lat2 - lat1)
    dlambda = math.radians(lon2 - lon1)
    a = math.sin(dphi / 2) ** 2 + math.cos(phi1) * math.cos(phi2) * math.sin(dlambda / 2) ** 2
    return 2 * R * math.asin(math.sqrt(a))

def build_graph_between_coords(coord1, coord2, buffer_km=50):
    # Create a bounding box that covers both points with a buffer
    mid_lat = (coord1[0] + coord2[0]) / 2
    mid_lon = (coord1[1] + coord2[1]) / 2

    max_dist = haversine_distance(coord1, coord2)
    buffer_dist = max_dist / 2 + buffer_km * 1000  # in meters

    # Filter for major roads only
    custom_filter = ('["highway"]["area"!~"yes"]'
                     '["highway"~"motorway|trunk|primary"]')

    G = ox.graph_from_point((mid_lat, mid_lon), dist=buffer_dist,
                            network_type='drive', custom_filter=custom_filter)
    return G




##############################################
# Functions for timing distance calculations #
##############################################

def test_osmnx(coord1, coord2):
    print(f"\nComputing route from {coord1} to {coord2}...")

    # Build the graph
    G = build_graph_between_coords(coord1, coord2)

    # Find nearest nodes
    orig_node = ox.distance.nearest_nodes(G, X=coord1[1], Y=coord1[0])
    dest_node = ox.distance.nearest_nodes(G, X=coord2[1], Y=coord2[0])

    try:
        # Compute shortest path length (in meters)
        length_m = nx.shortest_path_length(G, orig_node, dest_node, weight='length')  # Convert to kilometers
        length_km = length_m / 1000  # Convert to kilometers
        # print(f"OSMNx distance: {length_km:.2f} km")
        return length_km
    except nx.NetworkXNoPath:
        # print("No path found between these coordinates on the filtered road network.")
        return None

def test_google(coord1, coord2, mode="driving"):
    base_url = "https://maps.googleapis.com/maps/api/directions/json"
    params = {
        "origin": f"{coord1[0]},{coord1[1]}",
        "destination": f"{coord2[0]},{coord2[1]}",
        "mode": mode,
        "key": 'AIzaSyA25N_y-RSLDazUjvHPeL6Rh_dQgJCSfYY'
    }
    response = requests.get(base_url, params=params)
    data = response.json()

    if data['status'] != 'OK':
        print(f"Error from API: {data['status']}")
        return None

    route = data['routes'][0]['legs'][0]
    distance_meters = route['distance']['value']
    # print(f"Google Maps distance: {distance_meters / 1000:.2f} km")
    return distance_meters / 1000

def test_world_highways_scgraph(coord1, coord2):
    output = world_highways_geograph.get_shortest_path(
        origin_node={"latitude": coord1[0], "longitude": coord1[1]},
        destination_node={"latitude": coord2[0], "longitude": coord2[1]},
        output_units="km",
        off_graph_circuity=1.2
    )
    # print(f"Scgraph distance: {output['length']:.2f} km")
    return output['length']

def test_us_freeway_scgraph(coord1, coord2):
    output = us_freeway_geograph.get_shortest_path(
        origin_node={"latitude": coord1[0], "longitude": coord1[1]},
        destination_node={"latitude": coord2[0], "longitude": coord2[1]},
        output_units="km",
        off_graph_circuity=1.2
    )
    # print(f"Scgraph distance: {output['length']:.2f} km")
    return output['length']




cities = {
    'Los Angeles': (34.0522, -118.2437),
    'New York City': (40.7128, -74.0060),
    'Chicago': (41.8781, -87.6298),
    'Houston': (29.7604, -95.3698),
    'Phoenix': (33.4484, -112.0740),
    'Denver': (39.7392, -104.9903),
    'Seattle': (47.6062, -122.3321),
    'Miami': (25.7617, -80.1918),
    'Washington D.C.': (38.9072, -77.0369),
    'San Francisco': (37.7749, -122.4194),
    'Omaha': (41.2565, -95.9345),
    'Atlanta': (33.7490, -84.3880),
    'Austin': (30.2672, -97.7431),
    'Boston': (42.3601, -71.0589),
    'Las Vegas': (36.1699, -115.1398),
    'Detroit': (42.3314, -83.0458),
    'Kalamazoo': (42.2917, -85.5872),
    'Philadelphia': (39.952583, -75.165222),
    'San Diego': (32.7157, -117.1611),
    'Tampa': (27.9506, -82.4572),
    'Orlando': (28.5383, -81.3792),
}


# output = []

# For each pair of cities, calculate distances using all three methods
# cities_fully_visited = []
# print("\nDistance Calculations Between Cities:")
# for city1, coord1 in cities.items():
#     print(f"\nCalculating distances from {city1}...")
#     for city2, coord2 in cities.items():
#         if city1 != city2 and city2 not in cities_fully_visited:
#             output_item = {
#                 'city1': city1,
#                 'city2': city2,
#                 'coord1': coord1,
#                 'coord2': coord2
#             }
#             for function in [
#                 # Omitting Distance for OSMNx since it is too slow and requires too much data pulling
#                 # test_osmnx,
#                 test_world_highways_scgraph,
#                 test_us_freeway_scgraph,
#                 test_google
#             ]:
#                 time_start = time()
#                 length = function(coord1, coord2)
#                 elapsed_time = time() - time_start
#                 output_item[f"{function.__name__}_time_ms"] = elapsed_time * 1000
#                 output_item[f"{function.__name__}_length_km"] = length
#             output.append(output_item)
#             # print(f"{city1} to {city2}: {output_item}")
#     cities_fully_visited.append(city1)

# pamda.write_csv(
#     data=output,
#     filename="geo_time_comparison_tests.csv",
# )

```

```python
# ------------------------------------------------------------
# SCGraph-style Grid Path Visualisation (100×100 wall example)
# ------------------------------------------------------------
import matplotlib.pyplot as plt
from heapq import heappush, heappop

# ------------------------------------------------------------
# Build a 4‑neighbour grid graph with a wall at x = 50 (rows ≥ 5)
# ------------------------------------------------------------
def build_grid(x_size: int, y_size: int):
    blocks = {(x_size // 2, y) for y in range(5, y_size)}  # the wall
    nodes = {(x, y) for x in range(x_size) for y in range(y_size)
             if (x, y) not in blocks}

    # adjacency list: each cell connects to N, S, E, W neighbours
    adj = {u: [] for u in nodes}
    for x, y in nodes:
        for dx, dy in ((1, 0), (-1, 0), (0, 1), (0, -1)):
            v = (x + dx, y + dy)
            if v in nodes:
                adj[(x, y)].append(v)
    return nodes, blocks, adj

# ------------------------------------------------------------
# A* search with Manhattan (L₁) heuristic
# ------------------------------------------------------------
def a_star(adj, origin, dest):
    def h(u, v):
        return abs(u[0] - v[0]) + abs(u[1] - v[1])

    heap, dist, pred = [], {origin: 0}, {}
    heappush(heap, (h(origin, dest), 0, origin, None))

    while heap:
        _, g, u, parent = heappop(heap)
        if u in pred:                # already processed
            continue
        pred[u] = parent
        if u == dest:
            break
        for v in adj[u]:
            tentative = g + 1
            if v not in dist or tentative < dist[v]:
                dist[v] = tentative
                heappush(heap, (tentative + h(v, dest), tentative, v, u))

    # reconstruct
    if dest not in pred:
        return []                    # no path
    path, cur = [], dest
    while cur is not None:
        path.append(cur)
        cur = pred[cur]
    return path[::-1]

# ------------------------------------------------------------
# Plot helper
# ------------------------------------------------------------
def plot_case(ax, title, origin, dest, nodes, blocks, path):
    xs, ys = zip(*nodes)
    ax.scatter(xs, ys, s=1, alpha=0.15, label="Points")
    if blocks:
        bx, by = zip(*blocks)
        ax.scatter(bx, by, s=10, marker='s', c='black', alpha=1,
                   label="Wall")
    if path:
        px, py = zip(*path)
        ax.plot(px, py, lw=2, c='gold', label="Shortest Path")
        ax.scatter([origin[0], dest[0]],
                   [origin[1], dest[1]],
                   c='red', s=20, label="Start and End Points")
    ax.set_title(title)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

# ------------------------------------------------------------
# Main script
# ------------------------------------------------------------
if __name__ == "__main__":
    X, Y = 100, 100
    nodes, blocks, adj = build_grid(X, Y)

    cases = {
        "Bottom Left to Wall to Top Right (BLWBR)":
            ((5, 5),            (X - 5, Y - 5)),
        "Top Left to Wall to Top Right ()":
            ((5, Y - 5),        (X - 5, Y - 5)),
        "Up 5 Right 5":
            ((5, 5),            (10, 10)),
        "Bottom Left to Bottom Right":
            ((5, 5),            (X - 5, 5)),
    }

    fig, axes = plt.subplots(2, 2, figsize=(8, 8))
    for ax, (name, (orig, dest)) in zip(axes.flat, cases.items()):
        path = a_star(adj, orig, dest)
        plot_case(ax, name, orig, dest, nodes, blocks, path)

    handles, labels = axes[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="center", ncol=4,
               frameon=False, fontsize="small")
    plt.tight_layout()
    plt.show()
```

```python
REPEATS = 10
sc_times, sc_dists = [], []
ox_times, ox_dists = [], []
gm_times, gm_dists = [], []

for _ in range(REPEATS):
    # --- SCGraph ---
    t0 = time.perf_counter()
    sc_out = us_freeway_geograph.get_shortest_path(
        origin_node={"latitude": gm_origin[0], "longitude": gm_origin[1]},
        destination_node={"latitude": gm_dest[0],    "longitude": gm_dest[1]},
        output_units="km",
    )
    sc_times.append((time.perf_counter() - t0) * 1000)
    sc_dists.append(sc_out["length"])

    # --- OSMnx + NetworkX ---
    t0 = time.perf_counter()
    route = nx.shortest_path_length(G, nx_orig, nx_dest, weight="length")
    ox_times.append((time.perf_counter() - t0) * 1000)
    # convert meters → miles
    ox_dists.append(route)

    # --- Google Maps API ---
    t0 = time.perf_counter()
    directions = gmaps.directions(gm_origin, gm_dest, mode="driving")
    gm_times.append((time.perf_counter() - t0) * 1000)
    # Google returns meters
    gm_dists.append(directions[0]["legs"][0]["distance"]["value"]/1000)

# compute mean & std
def stats(xs):
    return np.mean(xs), np.std(xs)

sc_t_mean, sc_t_std = stats(sc_times)
ox_t_mean, ox_t_std = stats(ox_times)
gm_t_mean, gm_t_std = stats(gm_times)

sc_d_mean, sc_d_std = stats(sc_dists)
ox_d_mean, ox_d_std = stats(ox_dists)
gm_d_mean, gm_d_std = stats(gm_dists)

print(f"\nOver {REPEATS} runs:")
print(f"SCGraph          → Time: {sc_t_mean:.1f}±{sc_t_std:.1f} ms | Dist: {sc_d_mean:.2f}±{sc_d_std:.2f} km")
print(f"OSMNx + NetworkX → Time: {ox_t_mean:.1f}±{ox_t_std:.1f} ms | Dist: {ox_d_mean:.2f}±{ox_d_std:.2f} km")
print(f"Google Maps API  → Time: {gm_t_mean:.1f}±{gm_t_std:.1f} ms | Dist: {gm_d_mean:.2f}±{gm_d_std:.2f} km")
```
